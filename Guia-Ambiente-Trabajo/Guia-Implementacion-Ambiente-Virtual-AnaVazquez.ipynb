| Sistema | Aplicación recomendada |
|---|---|

## 6.1 Crear el ambiente

Ejecuta en la terminal:

```bash
conda create -n fisica-computacional python=3.10
```

Esta instrucción crea un nuevo ambiente Conda llamado `fisica-computacional` e instala **Python 3.10** dentro de él.

El comando puede interpretarse por partes:

```text
conda create
```

crea un nuevo ambiente;

```text
-n fisica-computacional
```

asigna al ambiente el nombre `fisica-computacional`;

```text
python=3.10
```

indica la versión de Python que se instalará.

En conjunto:

```text
Crear un ambiente
        ↓
fisica-computacional
        ↓
Python 3.10
```

Cuando aparezca:

```text
Proceed ([y]/n)?
```

Conda está solicitando confirmación para continuar con la instalación. Escribe:

```text
y
```

y presiona `Enter`.

Al finalizar, el ambiente estará creado. 

---

## 6.2 Activar el ambiente

Ejecuta en la terminal:

```bash
conda activate fisica-computacional
```

Esta instrucción **activa el ambiente Conda** que creamos anteriormente.

El comando puede interpretarse por partes:

```text
conda activate
```

indica a Conda que queremos comenzar a utilizar un ambiente;

```text
fisica-computacional
```

es el nombre del ambiente que deseamos activar.

En conjunto:

```text
Activar Conda
      ↓
fisica-computacional
      ↓
usar Python y las bibliotecas de este ambiente
```

Después de activarlo, el nombre del ambiente deberá aparecer al inicio del prompt:

```text
(fisica-computacional)
```

Por ejemplo:

```text
(fisica-computacional) usuario@equipo:~$
```

Esto indica que los comandos `python`, `pip` y las bibliotecas que instalemos se utilizarán dentro de este ambiente.

## 6.3 Verificar Python

Ejecuta en la terminal:

```bash
python --version
```

Esta instrucción muestra la **versión de Python** que está utilizando actualmente la terminal.

El comando puede interpretarse por partes:

```text
python
```

hace referencia al intérprete de Python activo;

```text
--version
```

solicita mostrar su número de versión.

Deberás obtener algo semejante a:

```text
Python 3.10.x
```

La `x` representa una versión de mantenimiento de Python 3.10, por ejemplo `3.10.12`.

> **Importante:** si aparece una versión diferente de Python, comprueba primero que el ambiente `fisica-computacional` esté activado.

---

## 6.4 Verificar la ruta de Python

Además de comprobar la versión, conviene verificar **qué instalación de Python está utilizando la terminal**.

### Ubuntu y macOS

Ejecuta:

```bash
which python
```

La instrucción `which` muestra la ubicación del ejecutable que se utilizará cuando escribamos `python`.

### Windows 11

Ejecuta en PowerShell:

```powershell
where.exe python
```

Esta instrucción muestra la ubicación del ejecutable de Python disponible en el entorno actual.

En ambos casos, la ruta debe corresponder al ambiente:

```text
fisica-computacional
```

Por ejemplo, puede aparecer una ruta semejante a:

```text
.../envs/fisica-computacional/bin/python
```

o, en Windows:

```text
...\\envs\\fisica-computacional\\python.exe
```

Esto confirma que estamos utilizando **Python 3.10 instalado dentro del ambiente del curso** y no otra instalación de Python presente en la computadora.

## 8.1 Registrar el ambiente como kernel de Jupyter

Después de instalar Jupyter, ejecuta:

```bash
python -m ipykernel install --user \
    --name fisica-computacional \
    --display-name "Python 3.10 (Física Computacional)"
```

Esta instrucción registra el ambiente `fisica-computacional` para que pueda seleccionarse como **kernel** dentro de Jupyter y Visual Studio Code.

Las partes principales son:

```text
python -m ipykernel install
```

ejecuta el módulo `ipykernel` y registra un nuevo kernel;

```text
--user
```

realiza el registro para el usuario actual;

```text
--name fisica-computacional
```

establece el nombre interno del kernel;

```text
--display-name "Python 3.10 (Física Computacional)"
```

define el nombre que veremos en la interfaz de Jupyter o VS Code.

En conjunto:

```text
ambiente fisica-computacional
          ↓
registrar como kernel
          ↓
Python 3.10 (Física Computacional)
          ↓
seleccionable en Jupyter y VS Code
```

### Windows 11

En PowerShell, escribe el mismo comando en una sola línea:

```powershell
python -m ipykernel install --user --name fisica-computacional --display-name "Python 3.10 (Física Computacional)"
```

> **Nota:** instalar Jupyter proporciona las herramientas para trabajar con notebooks; registrar `ipykernel` permite que esos notebooks utilicen específicamente el Python del ambiente `fisica-computacional`.


Las siguientes instrucciones comprueban que Python puede **importar correctamente las bibliotecas instaladas**.

```bash
python -c "..."
```

indica a Python que ejecute directamente el código escrito entre comillas, sin necesidad de crear un archivo `.py`.

## 10.1 Bibliotecas científicas

Ejecuta:

```bash
python -c "import numpy, scipy, matplotlib, pandas, sympy, sklearn; print('Stack científico: OK')"
```

La instrucción:

```text
import numpy, scipy, matplotlib, pandas, sympy, sklearn
```

intenta cargar las principales bibliotecas científicas que utilizaremos en el curso.

Si todas pueden importarse correctamente, Python ejecutará:

```text
print('Stack científico: OK')
```

y obtendrás:

```text
Stack científico: OK
```

El flujo de la comprobación es:

```text
Python
  ↓
importar bibliotecas
  ↓
¿se cargaron correctamente?
  ↓
Stack científico: OK
```

Si aparece un error como:

```text
ModuleNotFoundError
```

significa que alguna biblioteca no está disponible en el Python que se está utilizando.
