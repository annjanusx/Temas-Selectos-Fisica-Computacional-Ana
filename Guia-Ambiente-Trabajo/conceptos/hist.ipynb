# hist


## ¿Por qué `hist`?

Los histogramas son una herramienta fundamental en análisis experimental.

`hist` proporciona estructuras específicamente diseñadas para construir y manipular histogramas de forma más rica que una simple llamada de visualización.

La distinción es importante:

```text
Matplotlib → dibuja

hist → representa y manipula el histograma
```

Ambas herramientas pueden trabajar juntas.
