# Black


## ¿Por qué `black`?

Black aplica automáticamente un formato consistente al código Python.

Esto reduce discusiones sobre detalles puramente estéticos y permite concentrarnos en:

- estructura;
- algoritmo;
- corrección;
- interpretación física.

La idea es separar:

```text
cómo se ve el código
```

de

```text
qué hace el código
```
