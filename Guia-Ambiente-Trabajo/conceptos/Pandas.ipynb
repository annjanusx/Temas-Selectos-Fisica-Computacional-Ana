# Pandas


## ¿Por qué Pandas?

Pandas resulta útil cuando los datos tienen una estructura tabular:

```text
evento | energía | momento | masa | etiqueta
```

Proporciona herramientas cómodas para:

- leer datos;
- seleccionar columnas;
- filtrar registros;
- agrupar;
- ordenar;
- calcular resúmenes;
- exportar resultados.


## ¿Sustituye a NumPy?

No.

Una manera útil de pensarlo es:

```text
NumPy  → estructura numérica general
Pandas → datos tabulares con etiquetas
```

El curso utilizará cada herramienta cuando su representación sea natural para el problema.
