# Visual Studio Code


## ¿Qué es?

**Visual Studio Code (VS Code)** es el editor y entorno de desarrollo principal del curso.

Permite trabajar en un mismo lugar con:

- archivos `.py`;
- notebooks `.ipynb`;
- terminal;
- depuración;
- extensiones;
- Git;
- selección de intérpretes y kernels.


## ¿Por qué se eligió?

Porque permite introducir progresivamente herramientas profesionales sin abandonar un entorno accesible para principiantes.

Un estudiante puede comenzar con:

```text
archivo.py
```

y posteriormente incorporar:

```text
proyecto
├── módulos
├── pruebas
├── notebooks
├── datos
└── control de versiones
```

sin cambiar de aplicación principal.


## ¿Por qué no trabajar únicamente con JupyterLab?

JupyterLab es excelente para exploración interactiva. Sin embargo, el curso también pretende desarrollar hábitos de programación estructurada.

VS Code permite trabajar con notebooks y, al mismo tiempo, enseñar:

- scripts;
- módulos;
- funciones;
- clases;
- pruebas;
- organización de proyectos.

Por eso Jupyter se integra **dentro** del flujo de VS Code, en lugar de sustituirlo.
