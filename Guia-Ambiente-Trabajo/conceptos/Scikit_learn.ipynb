# Scikit-learn


## ¿Por qué Scikit-learn?

Para una introducción a *machine learning*, Scikit-learn proporciona una interfaz coherente para:

- preprocesamiento;
- clasificación;
- regresión;
- división entrenamiento/prueba;
- métricas;
- validación.

Su abstracción permite estudiar primero la lógica del análisis sin introducir inmediatamente infraestructuras más complejas de *deep learning*.


## ¿Por qué no comenzar directamente con redes neuronales?

Porque antes de estudiar modelos más complejos conviene comprender conceptos fundamentales como:

```text
variables de entrada
      ↓
entrenamiento
      ↓
predicción
      ↓
validación
      ↓
métricas
```

Scikit-learn es apropiado para esta primera capa conceptual.
