# Pylance


## ¿Por qué Pylance?

Pylance mejora la experiencia de edición en VS Code mediante:

- autocompletado;
- información sobre funciones;
- navegación entre símbolos;
- detección temprana de ciertos errores.

Es una ayuda para escribir código, no una biblioteca científica.
