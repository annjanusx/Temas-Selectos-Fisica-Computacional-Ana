# Terminal y shell


## ¿Son lo mismo?

En lenguaje cotidiano suelen confundirse, pero conceptualmente podemos distinguir:

```text
Terminal
   ↓
interfaz donde escribimos comandos

Shell
   ↓
programa que interpreta esos comandos
```

Ejemplos de shells:

```text
bash
zsh
PowerShell
```


## ¿Por qué utilizar la terminal?

Porque hace explícitas operaciones importantes:

```text
activar ambiente
cambiar de directorio
ejecutar programas
instalar paquetes
```

Aprender estas acciones proporciona una base transferible a servidores, clústeres y sistemas remotos.
