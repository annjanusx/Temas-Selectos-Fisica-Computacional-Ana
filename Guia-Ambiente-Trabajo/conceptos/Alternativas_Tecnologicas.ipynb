# Herramientas que deliberadamente no son el eje del curso


La existencia de una herramienta alternativa no implica que sea inadecuada.

Simplemente se priorizó un conjunto pequeño y coherente.

| Necesidad | Herramienta elegida | Algunas alternativas | Razón principal de la elección |
|---|---|---|---|
| Ambientes | Conda | `venv`, Poetry, uv | Administración multiplataforma y dependencias científicas |
| Editor | VS Code | PyCharm, Spyder | Scripts + notebooks + terminal + extensiones |
| Interactividad | Jupyter | IPython puro, notebooks web | Integración natural de texto, código y figuras |
| Arrays | NumPy | listas Python, JAX | Base estándar del stack científico |
| Métodos científicos | SciPy | implementaciones propias | Amplia colección de algoritmos |
| Visualización | Matplotlib | Plotly, Bokeh | Base general y muy extendida en ciencia |
| Tablas | Pandas | Polars | Interfaz pedagógica y ecosistema maduro |
| Símbolos | SymPy | sistemas CAS externos | Integración directa con Python |
| ML inicial | Scikit-learn | PyTorch, TensorFlow | Menor complejidad para conceptos iniciales |
| ROOT desde Python | uproot | ROOT/PyROOT | Integración ligera con arrays Python |
| Datos irregulares HEP | Awkward Array | listas anidadas | Operaciones vectorizadas sobre estructuras variables |

Las alternativas pueden estudiarse posteriormente cuando exista una necesidad científica concreta que justifique introducirlas.
