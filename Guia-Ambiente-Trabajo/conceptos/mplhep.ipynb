# mplhep


## ¿Por qué `mplhep`?

`mplhep` complementa Matplotlib con convenciones y estilos habituales en física de altas energías.

Su función no es reemplazar Matplotlib, sino especializar la presentación de ciertas figuras.

Podemos visualizar la relación como:

```text
Matplotlib
    ↓
base de visualización
    ↓
mplhep
    ↓
convenciones HEP
```
