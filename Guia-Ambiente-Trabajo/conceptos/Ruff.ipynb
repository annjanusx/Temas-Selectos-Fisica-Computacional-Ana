# Ruff


## ¿Por qué `ruff`?

Ruff analiza código Python para detectar problemas de estilo y diversos errores potenciales.

Su utilidad principal en el curso es ayudar al estudiante a adquirir hábitos de código legible y consistente.

No cambia la física del cálculo, pero reduce errores de programación y facilita revisar programas escritos por otras personas.
