# GitLens


## ¿Por qué aparece como opcional?

GitLens añade información visual sobre el historial de Git dentro de VS Code.

No es necesario para las primeras actividades del curso, por eso se mantiene opcional.

Puede ser útil cuando los estudiantes comiencen a trabajar con proyectos y control de versiones.
