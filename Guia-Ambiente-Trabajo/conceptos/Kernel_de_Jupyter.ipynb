# Kernel de Jupyter


## ¿Qué es?

El **kernel** es el proceso que ejecuta el código de las celdas de un notebook.

En este curso queremos que ese proceso corresponda al Python de:

```text
fisica-computacional
```

Por eso registramos:

```text
Python 3.10 (Física Computacional)
```


## Diferencia entre intérprete y kernel

De forma simplificada:

```text
archivo .py
   ↓
intérprete de Python
```

mientras que:

```text
notebook .ipynb
   ↓
kernel de Jupyter
   ↓
Python
```

Ambos deben apuntar al mismo ambiente para mantener una configuración coherente.
