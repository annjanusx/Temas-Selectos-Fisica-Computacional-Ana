# SymPy


## ¿Por qué SymPy?

SymPy permite trabajar con expresiones matemáticas simbólicas.

Por ejemplo:

- derivadas;
- integrales;
- simplificación algebraica;
- ecuaciones;
- matrices simbólicas.


## ¿Cuál es su función pedagógica?

Puede utilizarse para conectar una expresión analítica con su implementación computacional.

El objetivo no es sustituir el desarrollo matemático manual, sino utilizar el cálculo simbólico cuando ayude a:

- comprobar resultados;
- explorar expresiones;
- traducir fórmulas a código.
