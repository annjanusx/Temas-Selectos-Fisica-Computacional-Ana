# Awkward Array


## ¿Qué problema resuelve?

En física de partículas, distintos eventos pueden contener diferente cantidad de objetos.

Por ejemplo:

```text
evento 1 → 2 jets
evento 2 → 5 jets
evento 3 → 1 jet
```

Esto no tiene naturalmente la forma de una matriz rectangular ordinaria.

Awkward Array está diseñado para trabajar con estructuras irregulares manteniendo operaciones vectorizadas.


## ¿Por qué es importante?

Porque refleja de manera natural la estructura de datos por evento que aparece en análisis de colisiones.
