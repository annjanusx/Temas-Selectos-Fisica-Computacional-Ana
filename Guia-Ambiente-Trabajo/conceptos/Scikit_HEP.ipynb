# Scikit-HEP


## ¿Qué es?

Scikit-HEP es un ecosistema de herramientas Python orientadas a física de altas energías.

No debe imaginarse como una única biblioteca monolítica, sino como un conjunto de paquetes especializados que interoperan.

En este curso nos interesan especialmente:

```text
uproot
awkward
vector
hist
mplhep
particle
```


## ¿Por qué utilizar este ecosistema?

Porque permite aplicar herramientas modernas de Python a estructuras y problemas típicos de física de partículas.

Esto conecta el aprendizaje general de Python con problemas reales de HEP sin abandonar el lenguaje utilizado durante el resto del curso.
