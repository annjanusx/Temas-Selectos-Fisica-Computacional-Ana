# NumPy


## ¿Por qué NumPy?

NumPy proporciona arreglos multidimensionales y operaciones vectorizadas eficientes.

Muchas cantidades físicas se representan naturalmente como colecciones de valores:

```text
tiempos
energías
momentos
posiciones
mediciones
```

Con NumPy podemos expresar operaciones matemáticas de manera cercana a su forma conceptual.

Por ejemplo:


````python
E = np.sqrt(p**2 + m**2)
````

sobre un arreglo completo de momentos.


## Papel pedagógico

NumPy introduce una transición fundamental:

```text
calcular un valor
        ↓
calcular una colección de valores
        ↓
pensar en términos de datos científicos
```


## ¿Por qué es central?

Gran parte del ecosistema científico de Python está construido directa o indirectamente alrededor de arreglos NumPy.
