# uproot


## ¿Por qué uproot?

Los experimentos de física de altas energías utilizan ampliamente archivos ROOT.

`uproot` permite leer este tipo de datos desde Python y convertir su contenido a estructuras compatibles con el ecosistema científico de Python.

La idea conceptual es:

```text
archivo ROOT
     ↓
   uproot
     ↓
Python / NumPy / Awkward
```


## ¿Por qué es especialmente útil en este curso?

Porque queremos trabajar directamente en Python y evitar que el estudiante tenga que aprender simultáneamente otro entorno completo antes de poder analizar datos.
