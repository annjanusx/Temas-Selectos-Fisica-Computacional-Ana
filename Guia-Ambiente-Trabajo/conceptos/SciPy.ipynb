# SciPy


## ¿Por qué SciPy?

SciPy extiende NumPy con algoritmos científicos ya implementados y ampliamente utilizados.

Incluye herramientas para problemas como:

- integración numérica;
- optimización;
- interpolación;
- álgebra lineal;
- ecuaciones diferenciales;
- estadística.


## ¿Por qué no programar todos esos algoritmos desde cero?

En algunas actividades sí puede ser pedagógico implementar un algoritmo sencillo.

Sin embargo, después de comprender el método conviene aprender a utilizar implementaciones científicas robustas.

La secuencia didáctica ideal es:

```text
comprender el método
      ↓
implementar una versión sencilla
      ↓
utilizar una biblioteca científica
      ↓
comparar resultados
```
