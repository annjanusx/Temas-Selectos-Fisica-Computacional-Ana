# Matplotlib


## ¿Por qué Matplotlib?

La visualización es parte del razonamiento científico, no únicamente una etapa estética.

Matplotlib permite construir:

- curvas;
- histogramas;
- diagramas de dispersión;
- barras de error;
- figuras con varias capas de información.

En física computacional, una gráfica puede revelar rápidamente:

```text
tendencias
errores
regiones anómalas
comportamientos asintóticos
dependencias entre variables
```


## Papel en el curso

Se utilizará desde ejercicios elementales hasta visualizaciones de distribuciones de observables físicos.
