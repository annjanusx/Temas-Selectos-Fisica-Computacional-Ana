# Jupyter


## ¿Qué aporta un notebook?

Un notebook permite dividir el trabajo en celdas que combinan:

- código;
- ecuaciones;
- texto;
- resultados;
- gráficas.

Esto lo hace especialmente útil para enseñanza y exploración científica.

Por ejemplo:

```text
explicación física
      ↓
ecuación
      ↓
código
      ↓
gráfica
      ↓
interpretación
```

pueden aparecer de manera consecutiva dentro del mismo documento.


## ¿Por qué se utilizará junto con archivos `.py`?

Porque cumplen funciones complementarias.

### Notebook

Adecuado para:

- exploración;
- demostraciones;
- análisis interactivo;
- visualización;
- material docente.

### Script o módulo `.py`

Adecuado para:

- programas reutilizables;
- funciones;
- clases;
- pruebas;
- proyectos de mayor tamaño.

El curso utiliza ambos para evitar que el estudiante identifique “programar en Python” únicamente con trabajar en notebooks.
