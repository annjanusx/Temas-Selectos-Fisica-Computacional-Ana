# rich


## ¿Por qué `rich`?

`rich` permite presentar información estructurada en terminal:

- tablas;
- mensajes;
- indicadores;
- resaltado.

No es indispensable para realizar los cálculos físicos. Se incluye como herramienta auxiliar para mostrar cómo puede mejorarse la interacción con programas de línea de comandos.
