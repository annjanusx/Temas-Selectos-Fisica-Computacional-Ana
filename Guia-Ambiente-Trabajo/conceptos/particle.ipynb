# particle


## ¿Por qué `particle`?

Permite consultar propiedades de partículas mediante identificadores estándar.

Es útil para conectar la representación computacional con conceptos del Modelo Estándar:

```text
PDG ID
  ↓
partícula
  ↓
propiedades
```

Tiene además un valor pedagógico directo para actividades de programación orientada a objetos y manipulación de información física.
