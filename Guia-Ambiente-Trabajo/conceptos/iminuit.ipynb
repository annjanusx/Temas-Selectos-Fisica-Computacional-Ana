# iminuit


## ¿Por qué `iminuit`?

Muchos problemas de física requieren ajustar parámetros minimizando una función objetivo.

`iminuit` proporciona una interfaz Python para métodos de minimización ampliamente utilizados en análisis científico.

Puede utilizarse para introducir:

- funciones de costo;
- parámetros;
- mínimos;
- incertidumbres;
- ajustes a datos.


## ¿Por qué no utilizar solamente optimizadores genéricos?

SciPy seguirá siendo importante. `iminuit` se incorpora porque permite mostrar herramientas especialmente familiares en contextos de ajuste de parámetros en física.
