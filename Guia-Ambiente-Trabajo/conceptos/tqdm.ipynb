# tqdm


## ¿Por qué `tqdm`?

Algunos cálculos recorren miles o millones de elementos y pueden tardar un tiempo apreciable.

`tqdm` permite añadir una barra de progreso con muy poco código.

Esto mejora la interacción con programas largos y ayuda a distinguir entre:

```text
el programa sigue calculando
```

y

```text
el programa se detuvo
```
