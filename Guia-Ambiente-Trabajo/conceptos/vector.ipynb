# vector


## ¿Por qué `vector`?

En física de partículas aparecen continuamente:

- vectores espaciales;
- cuadrivectores;
- momento transversal;
- pseudorapidez;
- masa invariante.

La biblioteca `vector` proporciona objetos preparados para estas operaciones.

Esto permite escribir código más cercano al lenguaje físico:


```python
p.mass
p.pt
```

en lugar de reconstruir manualmente cada expresión en todos los ejercicios.


## Papel pedagógico

Primero puede desarrollarse explícitamente la matemática de un cuadrivector y después comparar esa implementación con una herramienta especializada.
