# Idea pedagógica final

La herramienta correcta no se selecciona porque sea popular, sino porque existe una correspondencia entre:

```text
estructura del problema
        ↓
representación matemática
        ↓
representación computacional
        ↓
biblioteca apropiada
```

Por ejemplo:

```text
muchos valores numéricos
        ↓
arrays
        ↓
NumPy
```

```text
eventos con distinto número de partículas
        ↓
datos irregulares
        ↓
Awkward Array
```

```text
cuadrivectores
        ↓
álgebra relativista
        ↓
vector
```

```text
archivo ROOT
        ↓
estructura de datos HEP
        ↓
uproot
```

Esta relación entre **problema físico y herramienta computacional** es más importante que memorizar comandos individuales.
