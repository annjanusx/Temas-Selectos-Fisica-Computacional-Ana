# Python 3.10


## ¿Por qué Python?

Python permite concentrarnos inicialmente en el problema físico sin introducir una cantidad excesiva de detalles de bajo nivel.

Su ecosistema científico hace posible utilizar el mismo lenguaje para:

- cálculo numérico;
- álgebra simbólica;
- visualización;
- análisis de datos;
- optimización;
- estadística;
- machine learning;
- análisis de datos de física de altas energías.

La idea pedagógica es mantener una continuidad:

```text
primer programa
      ↓
métodos numéricos
      ↓
datos experimentales
      ↓
análisis de física de partículas
```

sin cambiar continuamente de lenguaje.


## ¿Por qué fijar Python 3.10?

La elección de **Python 3.10** no significa que sea la versión más reciente ni que las versiones posteriores sean inadecuadas.

Se fija una versión concreta por razones de:

- reproducibilidad;
- compatibilidad;
- uniformidad entre los equipos;
- reducción de problemas durante una clase.

En docencia es preferible que todos ejecuten el mismo material sobre una base conocida antes que perseguir continuamente la versión más reciente.


## ¿Podría actualizarse en el futuro?

Sí. La versión del curso puede migrarse posteriormente después de verificar que notebooks, ejemplos y dependencias funcionan correctamente.

La idea importante es:

```text
versión fijada
≠
versión eterna
```
