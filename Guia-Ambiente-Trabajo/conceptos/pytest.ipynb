# pytest


## ¿Por qué introducir pruebas automáticas?

En computación científica un programa puede ejecutarse sin errores y, aun así, producir un resultado incorrecto.

Una prueba puede expresar una propiedad conocida:

```text
entrada conocida
      ↓
resultado esperado
```

y comprobar automáticamente que el código sigue satisfaciéndola.

`pytest` permite introducir esta práctica con una sintaxis relativamente sencilla.


## Relevancia científica

Las pruebas no sustituyen la validación física, pero ayudan a detectar errores de implementación antes de interpretar resultados.
