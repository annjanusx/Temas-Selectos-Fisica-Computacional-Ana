# uncertainties


## ¿Por qué `uncertainties`?

Permite representar cantidades acompañadas de incertidumbres y propagar errores en expresiones sencillas.

Su utilidad pedagógica está en conectar:

```text
medición
   +
incertidumbre
   ↓
resultado derivado
   +
incertidumbre propagada
```

No sustituye un análisis estadístico completo, pero resulta útil para introducir de forma computacional la propagación de errores.
